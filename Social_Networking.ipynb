```python
class SocialNetwork:
    def __init__(self, users):
        self.users = users
        self.n = len(users)

        # Create an adjacency matrix
        self.matrix = [
            [0] * self.n for _ in range(self.n)
        ]

    def add_friendship(self, user1, user2):
        i = self.users.index(user1)
        j = self.users.index(user2)

        self.matrix[i][j] = 1
        self.matrix[j][i] = 1

        print(user1, "and", user2, "are now friends.")

    def check_friendship(self, user1, user2):
        i = self.users.index(user1)
        j = self.users.index(user2)

        if self.matrix[i][j] == 1:
            print(user1, "and", user2, "are friends.")
        else:
            print(user1, "and", user2, "are not friends.")

    def display_matrix(self):
        print("\nAdjacency Matrix:")
        print("   ", *self.users)

        for i in range(self.n):
            print(self.users[i], self.matrix[i])


# Main program
users = ["Sanchi", "Rutuja", "Sanu", "Priti"]

network = SocialNetwork(users)

network.add_friendship("Sanchi", "Rutuja")
network.add_friendship("Sanchi", "Sanu")
network.add_friendship("Rutuja", "Priti")
network.add_friendship("Sanu", "Priti")

network.display_matrix()

network.check_friendship("Sanchi", "Rutuja")
network.check_friendship("Sanchi", "Priti")
```
